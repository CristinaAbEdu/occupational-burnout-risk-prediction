## Early Prediction of Occupational Burnout Risk Using Interpretable Machine Learning Models

### Purpose

This notebook documents the common methodological and computational framework used for the implementation phase of the study.

The study uses multiple independent secondary workplace datasets to investigate the early prediction of occupational burnout risk using interpretable machine learning. Each dataset will be analysed independently because the datasets differ in their populations, variables, measurement instruments and workplace contexts.

A consistent analytical framework will nevertheless be applied across the selected datasets to support methodological consistency and comparison of findings.

### Primary Datasets

The current primary datasets are:

1. OLBI Dataset
2. Burnout, Emotional Symptoms and Work Engagement Dataset (BAT-23)
3. Brazil Teacher Pandemic Burnout Dataset (MBI)
4. Workplace Incivility Dataset (EES)

### Predictive Models

The primary classification models are:

- Logistic Regression
- Decision Tree

These models were selected because they support supervised classification while maintaining a strong emphasis on interpretability.

### General Implementation Pipeline

1. Define and operationalise the burnout-risk outcome.
2. Inspect and preprocess each dataset.
3. Identify appropriate workplace-related predictor variables.
4. Prevent target leakage by separating burnout outcome variables from predictors.
5. Apply LLM-assisted semantic mapping to harmonise heterogeneous predictor constructs.
6. Conduct exploratory data analysis.
7. Prepare predictor (X) and outcome (y) variables for modelling.
8. Split data into training and testing sets.
9. Train Logistic Regression and Decision Tree models.
10. Apply cross-validation and model tuning where appropriate.
11. Evaluate model performance using classification metrics.
12. Interpret model outputs and identify influential workplace-related factors.
13. Compare recurring predictor constructs across independently analysed datasets.

### Important Methodological Principle

The datasets will not be merged into a single dataset. Predictive modelling will be performed independently for each dataset, while broader workplace constructs and model findings will subsequently be compared across datasets.

## Harmonised Burnout Score Framework

The selected primary datasets use different burnout or burnout-related
measurement instruments, including the OLBI, BAT-23, MBI and Emotional
Exhaustion Scale (EES). These instruments differ in their number of items,
response scales, scoring ranges and dimensional structure.

To support a consistent analytical framework across independently analysed
datasets, the study will transform the correctly calculated instrument-level
scores onto a common 0–100 metric using the Percent of Maximum Possible
(POMP) transformation.

The general transformation is:

HBS = ((S - S_min) / (S_max - S_min)) × 100

where:

- HBS = Harmonised Burnout Score
- S = participant's original burnout instrument score
- S_min = theoretical minimum possible score for the instrument
- S_max = theoretical maximum possible score for the instrument

The resulting Harmonised Burnout Score ranges from 0 to 100, with higher
values representing a position closer to the maximum possible burnout-related
score of the original instrument.

The transformation does not assume that the original burnout instruments are
psychometrically identical. Rather, it provides a consistent numerical metric
for representing scores originating from instruments with different scoring
ranges while preserving the participant's relative position within the
possible range of the original measure.

Instrument-specific scoring procedures will be applied before harmonisation.
Burnout questionnaire items used to construct the outcome will not be included
as predictor variables during supervised modelling in order to prevent target
leakage.

A binary burnout-risk outcome for classification will subsequently be derived
from the harmonised score using a separately defined and justified
operationalisation rule.

## Burnout-Risk Target Operationalisation

The Harmonised Burnout Score (HBS) will be retained as a continuous
0–100 measure for descriptive and exploratory analysis.

For supervised classification, a binary burnout-risk outcome will be
derived independently within each dataset.

The primary operationalisation will classify participants within the
upper quartile (75th percentile and above) of the HBS distribution as
having elevated burnout risk.

Burnout_Risk = 1: HBS >= dataset-specific 75th percentile
Burnout_Risk = 0: HBS < dataset-specific 75th percentile

This classification represents elevated burnout risk relative to the
respective study population and should not be interpreted as a clinical
diagnosis of burnout.

Because no universally validated clinical threshold exists for the
study-specific harmonised score, sensitivity analysis will subsequently
be used to examine whether the main modelling conclusions remain stable
under reasonable alternative risk thresholds.

The continuous HBS will be retained throughout the analysis to preserve
the original score information and support descriptive analysis.
