# Dataset Overview

Study: Burnout, Emotional Symptoms, and Work Engagement

Sample sizes:

- Psychologists dataset: 311 participants × 113 variables
- Workers dataset: 306 participants × 88 variables

Note: These sample sizes are lower than those originally anticipated in the research proposal.

Measurement Scales Used

According to the uploaded codebook, the dataset contains the following validated instruments:

# BAT-23 (Burnout Assessment Tool)

Measures occupational burnout across four dimensions:

Exhaustion (BAT1–BAT8)
Mental Distance (BAT9–BAT13)
Cognitive Impairment (BAT14–BAT18)
Emotional Impairment (BAT19–BAT23)

Response scale:

1 = Never
2 = Rarely
3 = Sometimes
4 = Often
5 = Always

# COPSOQ (Copenhagen Psychosocial Questionnaire)

Measures workplace demands and resources.

Demands

Quantitative demands (COP1–COP4)
Cognitive demands (COP5–COP8)
Emotional demands (COP9–COP11)
Demands to hide emotions (COP12–COP13)

Resources

Possibilities for development (COP14–COP17)
Social support (COP18–COP21)
Workplace relations (COP22–COP23)
Control (COP24–COP25)
UWES (Utrecht Work Engagement Scale)

Measures work engagement.

Dimensions include:

Vigor
Dedication
Absorption
PPFI (Personalized Psychological Flexibility Index)

Measures psychological flexibility through:

Avoidance
Acceptance
Harnessing
DASS-21

Measures emotional distress through:

Depression
Anxiety
Stress
AAQ-II

Measures:

Experiential avoidance
Why this dataset is promising

Note:

This dataset appears promising because it includes a validated burnout measure (BAT-23) as the potential target variable, together with workplace demands, workplace resources, work engagement, psychological flexibility, emotional symptoms, and demographic information that may serve as predictor variables in supervised machine learning models.
