```python
# Verify calculation of the Emotional Exhaustion score

data["EES_calculated"] = data[ees_columns].sum(axis=1)

comparison = data[
    ["EES", "KeletihanEmosi", "EES_calculated"]
].head(15)

display(comparison)

print(
    "\nEES matches calculated 9-item sum:",
    (data["EES"] == data["EES_calculated"]).all()
)
```

## Initial Dataset Assessment

**Decision: Retain as a strong candidate for primary modelling.**

### Dataset Characteristics
- 201 participant-level observations.
- 58 variables.
- No missing values were identified.
- No exact duplicate observations were identified.
- Includes demographic and occupational characteristics such as age, gender,
  education, employment sector, rank, tenure and work location.
- Contains documented measures of workplace incivility, emotional exhaustion
  and adaptive performance.

### Emotional Exhaustion Outcome
The dataset contains nine Emotional Exhaustion Scale (`EES`) items scored using
a 1–5 response scale.

The supplied `EES` variable was computationally verified as the sum of the nine
individual emotional exhaustion items.

Therefore:

`EES = EES.item1 + ... + EES.item9`

The observed scoring structure is consistent with a possible total score range
of 9–45.

The dataset therefore provides a transparent and traceable emotional-exhaustion
outcome that may be operationalised as a burnout-related risk target during the
later modelling stage.

No binary threshold was applied during dataset screening.

### Relevant Predictor Constructs
Potential predictors include:
- Workplace incivility
- Employment sector
- Employment rank
- Tenure
- Demographic characteristics

The dataset additionally contains an adaptive performance measure. Its role in
the final modelling framework will be considered during feature selection.

### Strengths
- Participant-level workplace data.
- Clearly documented questionnaire.
- Directly documented emotional exhaustion construct.
- Existing emotional exhaustion score can be independently reproduced from
  the individual questionnaire items.
- Contains workplace-related explanatory variables.
- No missing values or exact duplicates.

### Limitations
- Sample size is relatively modest (n = 201).
- Emotional exhaustion represents a burnout-related dimension rather than
  necessarily a complete multidimensional burnout construct.
- The final method for converting the emotional exhaustion score into the
  study's binary burnout-risk outcome has not yet been determined.

### Overall Suitability
This dataset is strongly aligned with the study because it combines a
documented emotional-exhaustion outcome with workplace-related explanatory
variables.

It will therefore be retained as a strong candidate for primary supervised
modelling, subject to the final cross-dataset selection and target-variable
operationalisation process.
