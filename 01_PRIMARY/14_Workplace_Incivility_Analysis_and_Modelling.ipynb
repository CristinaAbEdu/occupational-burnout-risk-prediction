## Burnout-Related Outcome Construction

Emotional exhaustion was used as the burnout-related outcome for this dataset. The supplied Emotional Exhaustion Scale (EES) consisted of nine items scored from 1 to 5 and was previously validated against the underlying questionnaire items for all participants.

To maintain a common outcome framework across the independently analysed datasets, the EES score was transformed to a 0–100 Harmonised Burnout Score (HBS) using the theoretical EES score range of 9 to 45.

The HBS provides a common interpretive scale across datasets while preserving the original scoring structure of the emotional-exhaustion measure.

```python
# Transform EES to the common 0-100 HBS scale

workplace["HBS"] = (
    (workplace["EES"] - 9) / (45 - 9)
) * 100

print("HBS summary:")
print(workplace["HBS"].describe())
```

### Elevated Burnout-Risk Classification

Consistent with the common modelling framework, the 75th percentile of the HBS distribution was used as the dataset-relative threshold for elevated burnout risk.

Participants with an HBS at or above the 75th-percentile threshold were classified as elevated risk (`1`), while participants below the threshold were classified as lower risk (`0`). This classification represents relative burnout-related risk within the dataset and should not be interpreted as a clinical diagnostic threshold.

```python
# Create the elevated burnout-risk target using the 75th percentile

hbs_threshold = workplace["HBS"].quantile(0.75)

workplace["Burnout_Risk"] = (
    workplace["HBS"] >= hbs_threshold
).astype(int)

print("HBS elevated-risk threshold:", hbs_threshold)

print("\nBurnout-risk counts:")
print(workplace["Burnout_Risk"].value_counts().sort_index())

print("\nBurnout-risk percentages:")
print(
    workplace["Burnout_Risk"]
    .value_counts(normalize=True)
    .sort_index()
    .mul(100)
    .round(1)
)
```

**Risk classification:**  
The 75th-percentile HBS threshold was 50.0. Using this dataset-relative threshold, 143 participants (71.1%) were classified as lower risk and 58 participants (28.9%) as elevated emotional-exhaustion/burnout risk. The elevated-risk proportion exceeded 25% because participants scoring exactly at the 75th-percentile threshold were included in the elevated-risk group.

### Assessment of Sparse Categorical Levels

Categorical predictors were assessed for sparsely represented levels before model development. Very small categories can produce unstable one-hot encoded features, particularly in a relatively small dataset. Category frequencies were therefore reviewed before determining whether conceptually related low-frequency levels should be grouped for modelling.

```python
# Identify categorical levels represented by fewer than 10 participants

for column in categorical_candidates:
    counts = workplace[column].value_counts()
    sparse = counts[counts < 10]

    if len(sparse) > 0:
        print(f"\n--- {column} ---")
        print(sparse)
```

**Sparse-category decision:**  
Sparse levels were reviewed individually rather than automatically grouped using a frequency threshold. Employment sectors represented by very small numbers of participants were combined into an `Other` category to reduce unstable one-hot encoded features. The two small pre-tertiary education categories were also combined into a broader pre-tertiary category.

Race was excluded from the primary modelling predictor set because the sample was overwhelmingly represented by one racial group, leaving insufficient observations across the remaining categories for meaningful model estimation. Marital status and tenure categories were retained because they represented distinct questionnaire-defined categories despite some smaller group sizes.

```python
# Group sparse employment-sector categories
main_sectors = ["Health", "Education", "Services", "Security"]

workplace["Employment_Sector_Grouped"] = (
    workplace["Employment_Sector_English"]
    .where(
        workplace["Employment_Sector_English"].isin(main_sectors),
        "Other"
    )
)

# Group the two small pre-tertiary education categories
education_mapping = {
    "STPM": "Pre-tertiary",
    "Pre-University/Matriculation/Certificate": "Pre-tertiary"
}

workplace["Education_Grouped"] = (
    workplace["Education_English"]
    .replace(education_mapping)
)

print("Employment sector:")
print(workplace["Employment_Sector_Grouped"].value_counts())

print("\nEducation:")
print(workplace["Education_Grouped"].value_counts())
```

## Final Predictor Set and Modelling Data

Following construct validation, exploratory analysis and assessment of sparse categorical levels, the final predictor set was defined for model development.

The predictor set included demographic and occupational characteristics, workplace incivility and adaptive performance. Grouped versions of education and employment sector were used to reduce sparse categorical levels. Race was excluded from the primary modelling set because the sample was overwhelmingly represented by one racial group, resulting in insufficient representation across the remaining categories.

Variables used to construct the emotional-exhaustion outcome, including the EES items, EES score and HBS, were excluded from the predictor matrix to prevent target leakage.

```python
# Define final predictor set and target

numeric_features = [
    "Age",
    "WIS",
    "API"
]

categorical_features = [
    "Gender_English",
    "Marital_Status_English",
    "Education_Grouped",
    "Employment_Sector_Grouped",
    "Rank_English",
    "Tenure_English",
    "Location_English"
]

final_predictors = numeric_features + categorical_features

X = workplace[final_predictors].copy()
y = workplace["Burnout_Risk"].copy()

print("X shape:", X.shape)
print("y shape:", y.shape)

print("\nFinal predictors:")
for predictor in final_predictors:
    print("-", predictor)

print("\nTarget distribution:")
print(y.value_counts().sort_index())
```

## Train-Test Split and Preprocessing

The dataset was divided into training and held-out test sets using an 80/20 split. Stratified sampling was applied to preserve the proportion of lower- and elevated-risk participants across both subsets.

A fixed random state was used to support reproducibility. All preprocessing and subsequent model fitting were performed using the training data, with the test set retained for final evaluation on previously unseen observations.

```python
from sklearn.model_selection import train_test_split

# Create reproducible stratified training and test sets

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    stratify=y,
    random_state=42
)

print("Training set:", X_train.shape)
print("Test set:", X_test.shape)

print("\nTraining target:")
print(y_train.value_counts().sort_index())

print("\nTest target:")
print(y_test.value_counts().sort_index())
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

# Preprocessing for numerical and categorical predictors

preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            StandardScaler(),
            numeric_features
        ),
        (
            "cat",
            OneHotEncoder(handle_unknown="ignore"),
            categorical_features
        )
    ]
)

print("Numerical features:", numeric_features)
print("Categorical features:", categorical_features)
```

## Baseline Model Development

Two interpretable supervised classification algorithms were initially evaluated: Logistic Regression and Decision Tree. These models were selected to provide complementary forms of interpretability, with Logistic Regression representing a linear probabilistic model and Decision Tree capturing potentially non-linear decision relationships.

The models were implemented within preprocessing pipelines to ensure that transformations were learned only from the relevant training data during model fitting and cross-validation. Class weighting was applied to account for the imbalance between the lower- and elevated-risk classes.

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier

# Logistic Regression pipeline
lr_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("model", LogisticRegression(
        class_weight="balanced",
        max_iter=1000,
        random_state=42
    ))
])

# Decision Tree pipeline
dt_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("model", DecisionTreeClassifier(
        class_weight="balanced",
        random_state=42
    ))
])

print("Baseline Logistic Regression and Decision Tree pipelines created.")
```

### Baseline Cross-Validation

Baseline model performance was assessed using 5-fold stratified cross-validation on the training data. Stratification preserved the relative distribution of the lower- and elevated-risk classes within each fold.

Accuracy, precision, recall, F1-score and ROC-AUC were evaluated. Particular attention was given to recall and F1-score for the elevated-risk class because the study aims to identify individuals showing elevated burnout-related risk while maintaining balanced predictive performance.

```python
from sklearn.model_selection import StratifiedKFold, cross_validate
import pandas as pd

# Define stratified 5-fold cross-validation
cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

# Evaluation metrics
scoring = {
    "Accuracy": "accuracy",
    "Precision": "precision",
    "Recall": "recall",
    "F1": "f1",
    "ROC_AUC": "roc_auc"
}

# Evaluate baseline models
models = {
    "Logistic Regression": lr_pipeline,
    "Decision Tree": dt_pipeline
}

baseline_results = []

for model_name, model in models.items():
    scores = cross_validate(
        model,
        X_train,
        y_train,
        cv=cv,
        scoring=scoring
    )

    baseline_results.append({
        "Model": model_name,
        "Accuracy": scores["test_Accuracy"].mean(),
        "Precision": scores["test_Precision"].mean(),
        "Recall": scores["test_Recall"].mean(),
        "F1": scores["test_F1"].mean(),
        "ROC_AUC": scores["test_ROC_AUC"].mean()
    })

baseline_cv_results = pd.DataFrame(baseline_results)

print("Baseline 5-Fold Cross-Validation Results:")
print(
    baseline_cv_results
    .set_index("Model")
    .round(3)
)
```

**Baseline performance:**  
Logistic Regression showed stronger baseline cross-validation performance than the Decision Tree, particularly for recall (0.578), F1-score (0.480) and ROC-AUC (0.695). The Decision Tree achieved similar accuracy and precision but identified fewer elevated-risk cases across the validation folds. These results represent untuned baseline performance and were used as a reference for subsequent hyperparameter optimisation.

## Hyperparameter Tuning

Hyperparameter tuning was conducted using stratified 5-fold cross-validation on the training data. F1-score for the elevated-risk class was used as the primary optimisation criterion because it balances precision and recall while remaining appropriate for the moderately imbalanced outcome.

The held-out test set was not used during hyperparameter selection.

```python
from sklearn.model_selection import GridSearchCV

# Hyperparameter grid for Logistic Regression
lr_param_grid = {
    "model__C": [0.01, 0.1, 1, 10, 100],
    "model__solver": ["liblinear", "lbfgs"]
}

lr_grid = GridSearchCV(
    estimator=lr_pipeline,
    param_grid=lr_param_grid,
    scoring="f1",
    cv=cv,
    n_jobs=-1
)

lr_grid.fit(X_train, y_train)

print("Best Logistic Regression parameters:")
print(lr_grid.best_params_)

print("\nBest cross-validation F1:")
print(round(lr_grid.best_score_, 3))
```

```python
# Hyperparameter grid for Decision Tree

dt_param_grid = {
    "model__max_depth": [3, 5, 7, None],
    "model__min_samples_split": [2, 5, 10],
    "model__min_samples_leaf": [1, 2, 5, 10]
}

dt_grid = GridSearchCV(
    estimator=dt_pipeline,
    param_grid=dt_param_grid,
    scoring="f1",
    cv=cv,
    n_jobs=-1
)

dt_grid.fit(X_train, y_train)

print("Best Decision Tree parameters:")
print(dt_grid.best_params_)

print("\nBest cross-validation F1:")
print(round(dt_grid.best_score_, 3))
```

### Tuned Model Cross-Validation

Following hyperparameter optimisation, the selected Logistic Regression and Decision Tree configurations were evaluated using the same stratified 5-fold cross-validation framework and performance metrics used for the baseline models.

This enabled comparison of model performance following optimisation before final evaluation on the held-out test set.

```python
# Retrieve the tuned model pipelines

tuned_lr = lr_grid.best_estimator_
tuned_dt = dt_grid.best_estimator_

tuned_models = {
    "Tuned Logistic Regression": tuned_lr,
    "Tuned Decision Tree": tuned_dt
}

tuned_results = []

for model_name, model in tuned_models.items():
    scores = cross_validate(
        model,
        X_train,
        y_train,
        cv=cv,
        scoring=scoring
    )

    tuned_results.append({
        "Model": model_name,
        "Accuracy": scores["test_Accuracy"].mean(),
        "Precision": scores["test_Precision"].mean(),
        "Recall": scores["test_Recall"].mean(),
        "F1": scores["test_F1"].mean(),
        "ROC_AUC": scores["test_ROC_AUC"].mean()
    })

tuned_cv_results = pd.DataFrame(tuned_results)

print("Tuned 5-Fold Cross-Validation Results:")
print(
    tuned_cv_results
    .set_index("Model")
    .round(3)
)
```

**Tuned model performance:**  
Hyperparameter optimisation improved the cross-validation performance of both models. Logistic Regression improved from an F1-score of 0.480 to 0.527 and from a ROC-AUC of 0.695 to 0.714. The Decision Tree showed a larger improvement, increasing from an F1-score of 0.409 to 0.590 and from a ROC-AUC of 0.586 to 0.758. The tuned Decision Tree also achieved a recall of 0.689, indicating improved identification of elevated-risk observations across the validation folds. Final model performance was subsequently assessed using the untouched held-out test set.

```python
# Combine baseline and tuned cross-validation results

cv_comparison = pd.concat(
    [baseline_cv_results, tuned_cv_results],
    ignore_index=True
)

cv_long = cv_comparison.melt(
    id_vars="Model",
    var_name="Metric",
    value_name="Score"
)

fig = px.bar(
    cv_long,
    x="Metric",
    y="Score",
    color="Model",
    barmode="group",
    text_auto=".2f",
    title="Baseline and Tuned Model Cross-Validation Performance"
)

fig.update_layout(
    template="plotly_white",
    title_x=0.5,
    height=580,
    yaxis_title="Mean Cross-Validation Score",
    yaxis_range=[0, 1],
    legend_title="Model"
)

fig.show()
```

## Final Held-Out Test Evaluation

Following hyperparameter tuning, the tuned Logistic Regression and Decision Tree models were evaluated on the held-out test set, which had not been used during model training or hyperparameter selection.

Performance was assessed using accuracy, precision, recall, F1-score and ROC-AUC. Confusion matrices were also examined to determine the number of correctly identified and missed elevated-risk participants. Particular attention was given to recall because failure to identify elevated-risk cases represents an important consideration within an early-warning modelling context.

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix
)

# Evaluate tuned models on the held-out test set

test_results = []

for model_name, model in tuned_models.items():

    # Fit the selected model on the complete training set
    model.fit(X_train, y_train)

    # Predictions for the untouched test set
    y_pred = model.predict(X_test)
    y_prob = model.predict_proba(X_test)[:, 1]

    # Confusion matrix
    tn, fp, fn, tp = confusion_matrix(y_test, y_pred).ravel()

    test_results.append({
        "Model": model_name,
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(y_test, y_pred),
        "Recall": recall_score(y_test, y_pred),
        "F1": f1_score(y_test, y_pred),
        "ROC_AUC": roc_auc_score(y_test, y_prob),
        "TN": tn,
        "FP": fp,
        "FN": fn,
        "TP": tp
    })

test_results_df = pd.DataFrame(test_results)

print("Held-Out Test Results:")
print(
    test_results_df
    .set_index("Model")
    .round(3)
)
```

### Extended Logistic Regression Optimisation

Following the initial tuning stage, an extended Logistic Regression optimisation was conducted using only the training data. The search examined different regularisation strengths and penalty types using stratified cross-validation.

This additional optimisation was performed without reference to the held-out test set to preserve its role as an independent evaluation sample.

```python
# Extended Logistic Regression hyperparameter search

extended_lr_param_grid = [
    {
        "model__C": [0.001, 0.005, 0.01, 0.05, 0.1, 0.5, 1, 5, 10],
        "model__penalty": ["l1", "l2"],
        "model__solver": ["liblinear"]
    }
]

extended_lr_grid = GridSearchCV(
    estimator=lr_pipeline,
    param_grid=extended_lr_param_grid,
    scoring="f1",
    cv=cv,
    n_jobs=-1
)

extended_lr_grid.fit(X_train, y_train)

print("Best extended Logistic Regression parameters:")
print(extended_lr_grid.best_params_)

print("\nBest cross-validation F1:")
print(round(extended_lr_grid.best_score_, 3))
```

```python
from sklearn.model_selection import cross_val_predict
from sklearn.metrics import precision_score, recall_score, f1_score
import numpy as np

# Use the extended tuned Logistic Regression
extended_tuned_lr = extended_lr_grid.best_estimator_

# Generate out-of-fold probabilities using training data only
oof_probabilities = cross_val_predict(
    extended_tuned_lr,
    X_train,
    y_train,
    cv=cv,
    method="predict_proba",
    n_jobs=-1
)[:, 1]

# Assess candidate probability thresholds
threshold_results = []

for threshold in np.arange(0.20, 0.81, 0.01):
    predictions = (oof_probabilities >= threshold).astype(int)

    threshold_results.append({
        "Threshold": threshold,
        "Precision": precision_score(y_train, predictions, zero_division=0),
        "Recall": recall_score(y_train, predictions, zero_division=0),
        "F1": f1_score(y_train, predictions, zero_division=0)
    })

threshold_results_df = pd.DataFrame(threshold_results)

best_threshold_row = threshold_results_df.loc[
    threshold_results_df["F1"].idxmax()
]

print("Best training-CV threshold:")
print(round(best_threshold_row["Threshold"], 2))

print("\nPerformance at selected threshold:")
print(best_threshold_row.round(3))
```

```python
# Fit the final Logistic Regression candidate on all training data
extended_tuned_lr.fit(X_train, y_train)

# Predict probabilities for the held-out test set
final_lr_prob = extended_tuned_lr.predict_proba(X_test)[:, 1]

# Apply the threshold selected using training CV only
final_lr_pred = (
    final_lr_prob >= best_threshold_row["Threshold"]
).astype(int)

# Calculate final held-out metrics
final_lr_accuracy = accuracy_score(y_test, final_lr_pred)
final_lr_precision = precision_score(y_test, final_lr_pred)
final_lr_recall = recall_score(y_test, final_lr_pred)
final_lr_f1 = f1_score(y_test, final_lr_pred)
final_lr_auc = roc_auc_score(y_test, final_lr_prob)

tn, fp, fn, tp = confusion_matrix(
    y_test,
    final_lr_pred
).ravel()

print("Extended Tuned Logistic Regression – Held-Out Test Results")
print("Accuracy:", round(final_lr_accuracy, 3))
print("Precision:", round(final_lr_precision, 3))
print("Recall:", round(final_lr_recall, 3))
print("F1:", round(final_lr_f1, 3))
print("ROC-AUC:", round(final_lr_auc, 3))

print("\nConfusion Matrix Counts:")
print("TN:", tn)
print("FP:", fp)
print("FN:", fn)
print("TP:", tp)
```

**Final Logistic Regression optimisation:**  
Extended optimisation of Logistic Regression selected L1 regularisation with C = 0.5 and the liblinear solver. Using out-of-fold predictions from the training data, a probability threshold of 0.49 was selected based on F1-score before final held-out evaluation.

On the held-out test set, the final Logistic Regression achieved an accuracy of 0.756, precision of 0.571, recall of 0.667, F1-score of 0.615 and ROC-AUC of 0.779. The model correctly identified 8 of the 12 elevated-risk participants and missed 4, while producing 6 false-positive classifications.

The improvement was achieved through optimisation conducted within the training data rather than adjustment based on held-out test performance. However, results should remain cautiously interpreted because the test set contained only 41 participants.

## Model Interpretability

Interpretability analysis was conducted to examine how the predictors contributed to model predictions. For Logistic Regression, model coefficients and corresponding odds ratios were extracted to examine the direction and relative magnitude of associations with elevated burnout-related risk.

For categorical predictors, coefficients represent the encoded category relative to the category omitted by the encoding process. Continuous predictors were standardised during preprocessing and should therefore be interpreted in relation to standardised changes rather than their original measurement units.

```python
# Extract transformed feature names and Logistic Regression coefficients

final_preprocessor = extended_tuned_lr.named_steps["preprocessor"]
final_lr_model = extended_tuned_lr.named_steps["model"]

feature_names = final_preprocessor.get_feature_names_out()
coefficients = final_lr_model.coef_[0]

lr_interpretation = pd.DataFrame({
    "Feature": feature_names,
    "Coefficient": coefficients
})

lr_interpretation["Odds_Ratio"] = np.exp(
    lr_interpretation["Coefficient"]
)

lr_interpretation["Absolute_Coefficient"] = (
    lr_interpretation["Coefficient"].abs()
)

lr_interpretation = lr_interpretation.sort_values(
    "Absolute_Coefficient",
    ascending=False
)

print(
    lr_interpretation[
        ["Feature", "Coefficient", "Odds_Ratio"]
    ].round(3)
)
```

### Logistic Regression Interpretation

The final Logistic Regression used L1 regularisation, which can shrink weak predictor coefficients exactly to zero. Non-zero coefficients were therefore examined to identify the encoded features retained by the fitted model.

Positive coefficients indicate an association with greater modelled odds of elevated risk, whereas negative coefficients indicate an association with lower modelled odds. Continuous predictors were standardised before model fitting; therefore, their coefficients and odds ratios correspond to a one-standard-deviation increase in the predictor.

Categorical coefficients are interpreted as contributions of the encoded categories within the fitted predictive model rather than conventional reference-category odds ratios because all categorical levels were retained during one-hot encoding.

```python
# Retain features with non-zero coefficients selected by L1 regularisation

lr_nonzero = lr_interpretation[
    lr_interpretation["Coefficient"].abs() > 1e-6
].copy()

lr_nonzero = lr_nonzero[
    ["Feature", "Coefficient", "Odds_Ratio"]
].sort_values(
    "Coefficient",
    ascending=False
)

print("Non-zero Logistic Regression coefficients:")
print(lr_nonzero.round(3))
```

```python
# Clean feature names for visualisation

lr_plot = lr_nonzero.copy()

lr_plot["Feature_Label"] = (
    lr_plot["Feature"]
    .str.replace("num__", "", regex=False)
    .str.replace("cat__", "", regex=False)
    .str.replace("_", " ", regex=False)
)

# Sort from negative to positive coefficient
lr_plot = lr_plot.sort_values("Coefficient")

fig = px.bar(
    lr_plot,
    x="Coefficient",
    y="Feature_Label",
    orientation="h",
    text="Coefficient",
    labels={
        "Coefficient": "Logistic Regression Coefficient",
        "Feature_Label": "Predictor"
    },
    title="Non-Zero Predictors Retained by Final Logistic Regression"
)

fig.update_traces(
    texttemplate="%{text:.2f}",
    textposition="outside"
)

fig.update_layout(
    template="plotly_white",
    title_x=0.5,
    height=600,
    yaxis_title="",
    xaxis_title="Coefficient"
)

fig.add_vline(
    x=0,
    line_dash="dash"
)

fig.show()
```

Because WIS was standardised before modelling, a one-standard-deviation increase in workplace incivility corresponded to approximately 2.48 times the modelled odds of elevated emotional-exhaustion risk, holding the other encoded features constant.

The remaining retained coefficients were substantially smaller. L1 regularisation also reduced several encoded feature coefficients to zero, indicating that they contributed little to the final linear decision function. The categorical coefficients are interpreted as contributions within the fitted predictive model rather than conventional reference-category comparisons.

## Dataset Summary and Conclusion

This dataset contained 201 participants and examined workplace incivility, emotional exhaustion and adaptive performance alongside demographic and occupational characteristics. Emotional exhaustion was used as the burnout-related outcome. The validated nine-item EES score was transformed to a 0–100 Harmonised Burnout Score (HBS), and the dataset-specific 75th-percentile threshold of 50.0 was used to identify elevated risk. This resulted in 143 participants (71.1%) being classified as lower risk and 58 (28.9%) as elevated risk.

Following language verification, construct validation, preprocessing and exploratory analysis, ten predictors were retained for primary modelling. Sparse employment-sector and education categories were grouped where conceptually appropriate, while race was excluded from the primary model because of highly uneven representation across categories.

#### Sparse Category Handling

Category frequencies were examined before fitting the inferential Logistic Regression model. The 16–20 years tenure category contained nine observations, all of which belonged to the lower-risk group, creating a risk of complete separation and unstable maximum-likelihood estimation.

For the inferential analysis only, this ordered tenure category was combined with the adjacent >20 years category to form a broader 16+ years group. The primary predictive modelling analysis remained unchanged.

```python
# Create inferential dataset with stable tenure grouping

workplace_inference = complete_inference.copy()

workplace_inference["Tenure_Inference"] = (
    workplace_inference["Tenure_English"]
    .replace({
        "16-20 years": "16+ years",
        ">20 years": "16+ years"
    })
)

print(
    pd.crosstab(
        workplace_inference["Tenure_Inference"],
        workplace_inference["Burnout_Risk"],
        margins=True
    )
)
```

```python
# Prepare final design matrix for inferential Logistic Regression

inferential_predictors = [
    "Age",
    "WIS",
    "API",
    "Gender_English",
    "Marital_Status_English",
    "Education_Grouped",
    "Employment_Sector_Grouped",
    "Rank_English",
    "Tenure_Inference",
    "Location_English"
]

X_workplace_inference = workplace_inference[
    inferential_predictors
].copy()

y_workplace_inference = workplace_inference[
    "Burnout_Risk"
].astype(int)

# Categorical predictors used in inferential model
categorical_inference = [
    "Gender_English",
    "Marital_Status_English",
    "Education_Grouped",
    "Employment_Sector_Grouped",
    "Rank_English",
    "Tenure_Inference",
    "Location_English"
]

# Dummy-code categorical predictors
X_workplace_inference = pd.get_dummies(
    X_workplace_inference,
    columns=categorical_inference,
    drop_first=True,
    dtype=float
)

# Ensure all predictors are numeric
X_workplace_inference = X_workplace_inference.astype(float)

print("Final inferential sample size:",
      len(X_workplace_inference))

print("Number of predictors after dummy coding:",
      X_workplace_inference.shape[1])

print("\nAny missing values:",
      X_workplace_inference.isna().sum().sum())

print("\nPredictor columns:")
for col in X_workplace_inference.columns:
    print("-", col)
```

```python
import statsmodels.api as sm

# Add intercept
X_workplace_inference_sm = sm.add_constant(
    X_workplace_inference
)

# Fit inferential Logistic Regression
workplace_inferential_lr = sm.Logit(
    y_workplace_inference,
    X_workplace_inference_sm
).fit(maxiter=100)

print(workplace_inferential_lr.summary())
```

```python
# Create inferential results table for Workplace Incivility dataset

workplace_ci = workplace_inferential_lr.conf_int()

workplace_inference_results = pd.DataFrame({
    "Predictor": workplace_inferential_lr.params.index,
    "Coefficient": workplace_inferential_lr.params.values,
    "Odds Ratio": np.exp(workplace_inferential_lr.params.values),
    "CI Lower": np.exp(workplace_ci[0].values),
    "CI Upper": np.exp(workplace_ci[1].values),
    "p-value": workplace_inferential_lr.pvalues.values
})

# Remove intercept
workplace_inference_results = workplace_inference_results[
    workplace_inference_results["Predictor"] != "const"
].copy()

workplace_inference_results["Significant (p < 0.05)"] = np.where(
    workplace_inference_results["p-value"] < 0.05,
    "Yes",
    "No"
)

workplace_inference_results["95% CI for OR"] = (
    workplace_inference_results["CI Lower"].map("{:.3f}".format)
    + " – "
    + workplace_inference_results["CI Upper"].map("{:.3f}".format)
)

workplace_inference_table = workplace_inference_results[
    [
        "Predictor",
        "Coefficient",
        "Odds Ratio",
        "95% CI for OR",
        "p-value",
        "Significant (p < 0.05)"
    ]
]

display(
    workplace_inference_table.style
    .format({
        "Coefficient": "{:.3f}",
        "Odds Ratio": "{:.3f}",
        "p-value": lambda x: "<0.001" if x < 0.001 else f"{x:.3f}"
    })
    .background_gradient(
        subset=["Odds Ratio"],
        cmap="Blues"
    )
    .set_caption(
        "Inferential Logistic Regression Results – Workplace Incivility Dataset"
    )
    .hide(axis="index")
)
```

#### Statistical Interpretation

The inferential Logistic Regression model was statistically significant overall (likelihood-ratio p < 0.001), indicating that the predictor set collectively showed evidence of association with elevated burnout risk.

Workplace incivility (WIS) was the only predictor demonstrating a statistically significant independent association with elevated burnout risk (β = 0.119, p < 0.001). The corresponding odds ratio indicates that each one-unit increase in WIS was associated with approximately 12.6% higher odds of elevated burnout risk, holding the remaining predictors constant.

The demographic, occupational and adaptive-performance predictors did not demonstrate statistically significant independent associations after multivariable adjustment. The 6–10 year tenure category approached, but did not meet, the predefined significance threshold (p = 0.074).

These findings represent adjusted statistical associations rather than causal effects. Statistical significance is interpreted as supplementary to the primary predictive modelling results and does not independently indicate predictive importance or model performance.
