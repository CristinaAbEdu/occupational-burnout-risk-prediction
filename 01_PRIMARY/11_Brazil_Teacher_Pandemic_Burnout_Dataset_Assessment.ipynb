## Initial Dataset Assessment

**Decision: Strong candidate for primary supervised modelling.**

### Strengths
- Contains participant-level data for 438 teachers.
- Includes 20 MBI burnout items and an existing MBI final score.
- Contains substantial occupational and workplace-related information.
- Relevant workplace factors include workload, working hours, organisational
  support, coworker support, institutional resources, training, technological
  working conditions and teaching experience.
- Includes an accompanying coding/data dictionary sheet.
- The burnout outcome is available for all 438 participants.
- The MBI final score provides an independent burnout-related outcome that may
  later be operationalised into a binary burnout-risk target.

### Considerations
- The sample size is smaller than originally anticipated but remains potentially
  suitable for modelling and will be assessed alongside the other selected datasets.
- The data were collected within the specific context of remote teaching during
  the COVID-19 pandemic, which may limit generalisability.
- Several quality-of-life and mental-state variables are present but do not need
  to be included as primary workplace predictors.
- Individual MBI items must not be included as predictors if the MBI final score
  is used to construct the target variable, to avoid target leakage.
- The supplied documentation does not specify a binary risk threshold for the
  MBI final score. Binary target construction will therefore be addressed during
  the preprocessing stage using the common target-operationalisation strategy
  adopted across the final selected datasets.

### Overall Suitability
The dataset is highly relevant to the study because it combines an occupational
burnout measure with multiple workplace-related factors at participant level.
It should be retained as a strong candidate for primary modelling.
