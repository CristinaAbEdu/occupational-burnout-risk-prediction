Dataset Evaluation – Synthetic Workplace Burnout Dataset
Original Purpose

This dataset appears to have been created to simulate workplace conditions and burnout-related factors for machine learning and data analysis purposes. It contains numerous workplace variables commonly associated with occupational burnout.

Strengths
Includes a wide range of workplace-related variables relevant to occupational burnout, such as:
Work-life balance
Weekly working hours
Manager support
Remote work
Department
Years at company
Mental health days
Therapy access
Salary range
Burnout level
Burnout risk
The variables are well structured and suitable for machine learning modelling.
Requires relatively little preprocessing because the variables are already organised and labelled.
Limitations

The dataset appears to be synthetically generated rather than collected from real employees.

As a result:

the relationships between variables are simulated rather than observed,
the burnout risk and burnout level variables are artificially generated,
the dataset does not reflect authentic workplace experiences or employee responses.

Since this study aims to analyse real-world secondary datasets representing working individuals, the use of synthetic data would not align with the methodological approach described in the research proposal.

Alignment with Research Aim

Although the variables closely match the concepts investigated in this study, the dataset does not satisfy the requirement for analysing real employee data. Consequently, it would not provide evidence that reflects actual workplace burnout experiences.

Preliminary Decision

Rejected as a primary research dataset.

However, the dataset may be valuable for:

testing preprocessing pipelines,
validating machine learning code,
comparing the behaviour of models trained on synthetic versus real-world datasets,
demonstrating methodological differences between synthetic and real datasets if discussed as a limitation or future research direction.
